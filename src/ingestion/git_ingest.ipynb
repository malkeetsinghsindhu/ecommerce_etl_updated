```python
import pandas as pd
# Pandas used for data manipulation and data analysis
from pyspark.sql import SparkSession 
#Pascal Case also known as Upper Camer Case
```

```python
# This uses pandas to read a CSV file directly from a URL 
df=pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/customers.csv")

#This creates (or retrieves) a SparkSession, which is your entry point to all Spark functionality.
spark=SparkSession.builder.getOrCreate()   

#This Creates Pandas dataframe to Spark Dataframe 

spark_df= spark.createDataFrame(df)

#This writes the Spark DataFrame to storage

spark_df.write.format("csv").mode("overwrite").save("/Volumes/ecommerce_etl_updated/bronze/raw_data/customers")



```
